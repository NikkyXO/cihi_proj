```python
from pathlib import Path
import pandas as pd
```

```python
print (Path.cwd())
```

```python
BASE_DIR = next(
    (path for path in (Path.cwd(), *Path.cwd().parents)
     if (path / 'requirements.txt').is_file()),
    None,
)

print(BASE_DIR)
if BASE_DIR is None:
    raise FileNotFoundError(f'Could not find requirements.txt from {Path.cwd()}')
print(BASE_DIR)
```

```python
RAW_DATA_DIR = BASE_DIR / 'data' / 'raw' / 'nacrs-ed-2025-2026.xlsx'
CLEAN = BASE_DIR / 'data' / 'clean'
CLEAN.mkdir(parents=True, exist_ok=True)

```

```python

table1 = pd.read_excel(RAW_DATA_DIR, sheet_name='Table 1', header=1, nrows=10)
```

```python
# print(table1.to_string())
print(table1.info())
```

```python
print(table1.to_string())
print(table1.columns.to_list())
```

```python
table1.columns = [
    "province", "total_visits", "ctas_1_3_discharged", "ctas_4_5_discharged",
    "admitted", "median_ctas_1_3", "median_ctas_4_5", "median_admitted",
    "p90_ctas_1_3", "p90_ctas_4_5", "p90_admitted",
]
newTable = table1.dropna(subset=["total_visits"])

print(newTable.info())
print(newTable.to_string())
```

// save the cleaned data to a CSV file

```python

newTable.to_csv(CLEAN / 'cleaned_data.csv', index=False)
print(newTable.shape)
```

```python
print(newTable[["province", "total_visits", "admitted", "median_admitted", "p90_admitted"]].to_string(index=False))
```

```python
table2 = pd.read_excel(RAW_DATA_DIR, sheet_name='Table 2', header=1)
print(table2.info())
print(table2.columns.to_list())
table2.columns = [
    "province", "month", "sex", "age_0_4", "age_5_19", "age_20_64", "age_65_plus", "age_all"
]
print(table2.columns.to_list())
```

```python
table2["month"] = table2["month"].ffill()
table2["province"] = table2["province"].ffill()
table2 = table2.dropna(subset=["sex"])
table2= table2[table2["sex"].isin(["Male", "Female", "All"])]
table2 = table2.dropna(subset=["age_all"])
table2.to_csv(CLEAN / "table2_month_age_sex.csv", index=False)
print(table2.shape)

print(table2.info())
```

```python
print("Provinces in Table 2:", sorted(table2["province"].dropna().unique()))
print("Ontario rows", (table2["province"] == "Ont.").sum())
```
